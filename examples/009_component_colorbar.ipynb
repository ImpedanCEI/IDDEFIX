# Component plot for many resonators

This quick synthetic example exercises `plot_model_components` without fitting or external data. Set `N_RESONATORS` below 10 to see the usual legend; set it to 10 or more to see the shared, numbered colorbar. Matplotlib's font size can be adjusted as usual if needed.

```python
import matplotlib.pyplot as plt
import numpy as np
import iddefix

N_RESONATORS = 15  # Try 9, 10, or 20.
frequency = np.linspace(0.1e9, 2.5e9, 1200)
parameters = np.column_stack(
    (
        np.linspace(60, 140, N_RESONATORS),
        np.linspace(15, 30, N_RESONATORS),
        np.linspace(0.35e9, 2.25e9, N_RESONATORS),
    )
)
impedance = iddefix.Impedances.n_Resonator_longitudinal_imp(
    frequency, parameters.ravel()
)
parameter_bounds = [
    bound
    for rs, q, fr in parameters
    for bound in ((rs / 2, rs * 2), (q / 2, q * 2), (fr * 0.9, fr * 1.1))
]
model = iddefix.EvolutionaryAlgorithm(
    frequency, impedance, N_RESONATORS, parameter_bounds
)
model.evolutionParameters = parameters.ravel()  # Known synthetic parameters.
```

```python
fig, axes = model.plot_model_components(cmap="turbo", figsize=(11, 8.5), dpi=110)
plt.close(fig)
```
